```python
#Import Libraries & Load Dataset
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Load dataset
data = pd.read_csv("Dataset .csv")

# Preview relevant columns
data[['Restaurant Name', 'Votes', 'Aggregate rating']].head()
```

```python
#Ensure Votes and Ratings Are Numeric
```

```python
# Convert Votes and Aggregate rating to numeric (if needed)
data['Votes'] = pd.to_numeric(data['Votes'], errors='coerce')
data['Aggregate rating'] = pd.to_numeric(data['Aggregate rating'], errors='coerce')

# Drop rows with missing Votes or Ratings
data_clean = data.dropna(subset=['Votes', 'Aggregate rating'])
```

```python
#Identify Restaurants with Highest and Lowest Votes
```

```python
# Highest votes
highest_votes = data_clean.loc[data_clean['Votes'].idxmax()]
print("Restaurant with highest votes:")
print(highest_votes[['Restaurant Name', 'Votes', 'Aggregate rating']])

# Lowest votes
lowest_votes = data_clean.loc[data_clean['Votes'].idxmin()]
print("\nRestaurant with lowest votes:")
print(lowest_votes[['Restaurant Name', 'Votes', 'Aggregate rating']])
```

```python
#Distribution of Votes
```

```python
plt.figure(figsize=(8,5))
sns.histplot(data_clean['Votes'], bins=30, kde=True, color='skyblue')
plt.title("Distribution of Votes Across Restaurants")
plt.xlabel("Number of Votes")
plt.ylabel("Count of Restaurants")
plt.show()
```

```python
#Correlation Between Votes and Ratings
```

```python
plt.figure(figsize=(8,6))
sns.scatterplot(x='Votes', y='Aggregate rating', data=data_clean, alpha=0.5)
plt.title("Votes vs Aggregate Rating")
plt.xlabel("Number of Votes")
plt.ylabel("Aggregate Rating")
plt.show()

# Calculate correlation
corr = data_clean[['Votes','Aggregate rating']].corr().iloc[0,1]
print(f"Correlation between Votes and Ratings: {corr:.2f}")
```

```python
#Summary of Insights
```

```python
print("Votes Analysis Summary:")
print(f"• Restaurant with highest votes: {highest_votes['Restaurant Name']} ({highest_votes['Votes']} votes)")
print(f"• Restaurant with lowest votes: {lowest_votes['Restaurant Name']} ({lowest_votes['Votes']} votes)")
print(f"• Correlation between votes and rating: {corr:.2f}")
if corr > 0:
    print("• Restaurants with more votes generally have higher ratings.")
elif corr < 0:
    print("• Restaurants with more votes generally have lower ratings.")
else:
    print("• No strong relationship between votes and rating.")
```
