```python
#Import Libraries & Load Dataset
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Load dataset
data = pd.read_csv("Dataset .csv")

# Display first few rows
data.head()
```

```python
#Analyze Rating Text Distribution
```

```python
rating_counts = data['Rating text'].value_counts()
print(rating_counts)

plt.figure(figsize=(8,5))
sns.barplot(x=rating_counts.index, y=rating_counts.values, palette='viridis')
plt.title("Distribution of Rating Text")
plt.xlabel("Rating Text")
plt.ylabel("Number of Restaurants")
plt.xticks(rotation=45)
plt.show()
```

```python
#Positive vs Negative Ratings
```

```python
# Define positive and negative
positive_ratings = ['Excellent', 'Very Good']
negative_ratings = ['Poor', 'Average']

pos_count = data[data['Rating text'].isin(positive_ratings)].shape[0]
neg_count = data[data['Rating text'].isin(negative_ratings)].shape[0]
neutral_count = data.shape[0] - (pos_count + neg_count)

print(f"Positive: {pos_count}, Neutral: {neutral_count}, Negative: {neg_count}")

# Pie chart
plt.figure(figsize=(6,6))
plt.pie([pos_count, neutral_count, neg_count], labels=['Positive','Neutral','Negative'], 
        autopct='%1.1f%%', colors=['green','grey','red'])
plt.title("Rating Sentiment Distribution")
plt.show()
```

```python
#Average Aggregate Rating per Sentiment
```

```python
avg_positive = data[data['Rating text'].isin(positive_ratings)]['Aggregate rating'].mean()
avg_negative = data[data['Rating text'].isin(negative_ratings)]['Aggregate rating'].mean()

print(f"Average rating for positive reviews: {avg_positive:.2f}")
print(f"Average rating for negative reviews: {avg_negative:.2f}")
```
