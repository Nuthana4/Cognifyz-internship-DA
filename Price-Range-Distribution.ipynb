```python
#Import Libraries & Load Dataset
```

```python
import pandas as pd
import matplotlib.pyplot as plt

# Load dataset
data = pd.read_csv("Dataset .csv")

# Preview column
print(data['Price range'].head())
```

```python
#Check for Missing or Unusual Values
```

```python
print("Unique Price Range Values:", data['Price range'].unique())
print("Missing Values:", data['Price range'].isnull().sum())
```

```python
#Count Restaurants per Price Range
```

```python
price_counts = data['Price range'].value_counts().sort_index()
print(price_counts)
```

```python
#Calculate Percentages
```

```python
total_restaurants = len(data)
price_percentage = (price_counts / total_restaurants) * 100
print("\nPercentage of restaurants in each price range:")
print(price_percentage.round(2))
```

```python
#Bar Chart (Clear & Simple)  How many restaurants fall under each price category?
```

```python
plt.figure(figsize=(7,5))
plt.bar(price_counts.index, price_counts.values, color='skyblue', edgecolor='black')
plt.title('Distribution of Price Ranges Among Restaurants')
plt.xlabel('Price Range')
plt.ylabel('Number of Restaurants')
plt.xticks(price_counts.index)
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.show()
```

```python
#Pie Chart (Great for Percentages)  Proportion of restaurants in each price range (1–4).
```

```python
plt.figure(figsize=(6,6))
plt.pie(price_percentage, labels=price_percentage.index,
        autopct='%1.1f%%', startangle=140, colors=['lightgreen','skyblue','orange','pink'])
plt.title('Percentage of Restaurants by Price Range')
plt.show()
```

```python
#Combine Both Results in One Table
```

```python
price_df = pd.DataFrame({
    'Price Range': price_counts.index,
    'Restaurant Count': price_counts.values,
    'Percentage': price_percentage.values.round(2)
})
print(price_df)
```
