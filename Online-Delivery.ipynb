```python
#Import Libraries & Load Data
```

```python
import pandas as pd
import matplotlib.pyplot as plt

# Load dataset
data = pd.read_csv("Dataset .csv")

# Check relevant columns
print(data[['Has Online delivery', 'Aggregate rating']].head())
```

```python
#Check Unique Values
```

```python
print("Unique values in 'Has Online delivery':")
print(data['Has Online delivery'].unique())
```

```python
#Percentage of Restaurants Offering Online Delivery
```

```python
# Count total and 'Yes' responses
online_delivery_counts = data['Has Online delivery'].value_counts()

total_restaurants = len(data)
online_yes = online_delivery_counts.get('Yes', 0)
online_no = online_delivery_counts.get('No', 0)

# Calculate percentages
online_yes_percent = (online_yes / total_restaurants) * 100
online_no_percent = (online_no / total_restaurants) * 100

print(f" Percentage of restaurants with online delivery: {online_yes_percent:.2f}%")
print(f" Percentage of restaurants without online delivery: {online_no_percent:.2f}%")
```

```python
#Compare Average Ratings
```

```python
# Average rating of restaurants with online delivery
avg_rating_yes = data[data['Has Online delivery'] == 'Yes']['Aggregate rating'].mean()

# Average rating of restaurants without online delivery
avg_rating_no = data[data['Has Online delivery'] == 'No']['Aggregate rating'].mean()

print(f"Average rating (with online delivery): {avg_rating_yes:.2f}")
print(f"Average rating (without online delivery): {avg_rating_no:.2f}")
```

```python
#Pie Chart for Delivery Availability
```

```python
plt.figure(figsize=(6,6))
plt.pie([online_yes_percent, online_no_percent],
        labels=['Online Delivery', 'No Online Delivery'],
        autopct='%1.1f%%',
        colors=['lightgreen', 'lightcoral'],
        startangle=140)
plt.title('Percentage of Restaurants Offering Online Delivery')
plt.show()
```

```python
#Bar Chart Comparing Ratings
```

```python
plt.figure(figsize=(6,4))
plt.bar(['With Online Delivery', 'Without Online Delivery'],
        [avg_rating_yes, avg_rating_no],
        color=['skyblue', 'orange'])
plt.title('Average Ratings Comparison')
plt.ylabel('Average Rating')
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.show()
```
