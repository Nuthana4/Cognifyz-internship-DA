```python
#Import Libraries & Load Dataset
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Load dataset
data = pd.read_csv("Dataset .csv")

# Check relevant columns
data[['City', 'Longitude', 'Latitude']].head()
```

```python
#Clean & Prepare Data
```

```python
# Remove rows with missing coordinates
data = data.dropna(subset=['Longitude', 'Latitude'])

# Check number of restaurants and cities
print(f"Total restaurants: {len(data)}")
print(f"Total cities: {data['City'].nunique()}")
```

```python
#Basic Scatter Plot (All Restaurants)
```

```python
plt.figure(figsize=(8,6))
plt.scatter(data['Longitude'], data['Latitude'], alpha=0.4, s=20, color='dodgerblue')
plt.title('Restaurant Locations (All Cities)')
plt.xlabel('Longitude')
plt.ylabel('Latitude')
plt.grid(True, linestyle='--', alpha=0.6)
plt.show()
```

```python
#City-Wise Distribution Visualization
```

```python
plt.figure(figsize=(10,6))
sns.scatterplot(data=data, x='Longitude', y='Latitude', hue='City', s=50, alpha=0.6)
plt.title('Restaurant Locations by City')
plt.xlabel('Longitude')
plt.ylabel('Latitude')
plt.legend(bbox_to_anchor=(1.05, 1), loc='upper left')
plt.show()
```

```python
#Geographic Heatmap (Optional – Using Seaborn)
```

```python
plt.figure(figsize=(8,6))
sns.kdeplot(
    x=data['Longitude'],
    y=data['Latitude'],
    fill=True,
    cmap='coolwarm',
    bw_adjust=0.5,
    alpha=0.8
)
plt.title('Restaurant Density Heatmap')
plt.xlabel('Longitude')
plt.ylabel('Latitude')
plt.show()
```

```python
#Identify Top Cities with Most Restaurants
```

```python
city_counts = data['City'].value_counts().head(10)

print("Top 10 Cities with the Most Restaurants:")
print(city_counts)

plt.figure(figsize=(8,4))
city_counts.sort_values().plot(kind='barh', color='lightgreen', edgecolor='black')
plt.title('Top 10 Cities with Most Restaurants')
plt.xlabel('Number of Restaurants')
plt.ylabel('City')
plt.grid(axis='x', linestyle='--', alpha=0.7)
plt.show()
```

```python
#Summary & Insights
```

```python
print("Geographic Insights:")
print("• Areas with dense clusters have many restaurants — possibly metro or urban centers.")
print("• Sparse points represent suburban or rural areas with fewer dining options.")
print("• The top city (from the bar chart) likely shows the most visible cluster in the scatter plot.")
```
